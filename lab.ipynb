```python
import pandas as pd
import numpy as np
```

```python

data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'A'],
    'Value': [10, 15, 20, 25, 30, 35, 40, 50],
    'Fee': [2000, 25000, 26000, 22000, 24000, 35000, 40000, 5600],
    'Duration': ['30day', '40days', '35days', '40days', '60days', '60days', '50days', '60days'],
    'Discount': [1000, 2300, 1200, 2500, 2000, 500, 600, 1000]
}

df = pd.DataFrame(data)
print(df.head())
```

```python
df['Value'].agg('cumsum')

df[['Value']].cumsum()
```

```python
# Find the total, mean, max, and count of the Value column

total_sum = df['Value'].sum()
mean_value = df['Value'].mean()
max_value = df['Value'].max()
count_value = df['Value'].count()

print(f"Total sum: {total_sum}")
print(f"Mean: {mean_value}")
print(f"Max: {max_value}")
print(f"Count: {count_value}")
```

```python
result = df['Value'].agg(['sum', 'mean', 'max', 'count'])
print(result)
```

```python
# Mean
mean1 = df['Value'].agg('mean')
print(f'Mean: {mean1}')

mean2 = df[['Value']].mean()
print(f'Mean: {mean2}')
```

```python
# Max
max1 = df['Value'].agg('max')
print(max1)

max2 = df[['Value']].max()
print(max2)
```

```python
# Count
count1 = df['Value'].agg('count')
print(count1)

count2 = df[['Value']].count()
print(count2)
```

```python
result = df[['Fee', 'Discount']].agg(['min', 'max', 'mean'])
result
```
